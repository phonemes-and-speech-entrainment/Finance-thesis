```python
import yfinance as yf
import pandas as pd
```

```python
# Stock tickers to download data
tickers = ["AAPL", "MSFT", "NVDA", "AMZN", "META", "GOOGL", "TSLA"]

# Download 10 years of daily data, which will be resampled - close prices only
data = yf.download(tickers, period="10y", interval="1d", auto_adjust=False)["Adj Close"]
```

```python
# Function to calculate returns for a given period (week/month/year), 
# using pct_change ---> try with formula instead
def compute_returns(data, freq):
    resampled = data.resample(freq).last()
    returns = resampled.pct_change().dropna()
    return returns
```

```python
# Calculate returns
weekly_returns = compute_returns(data, 'W')   # Weekly frequency
monthly_returns = compute_returns(data, 'M')  # Monthly frequency
yearly_returns = compute_returns(data, 'Y')   # Yearly frequency
```

```python
#print all weekly returns
print(weekly_returns.to_string())
```

```python
#print all monthly returns
print(monthly_returns.to_string())
```

```python
#print all yearly returns
print(yearly_returns.to_string())
```
